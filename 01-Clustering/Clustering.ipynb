## Neuron Implementation

```python
import torch
import torch.nn as nn
```

```python
# input 

torch.manual_seed(42)
inputs = torch.tensor([1.0, 2.0, 3.0])    #(x1, x2, x3) --> 1D array/Vector , Tensors are multidimensional Array
```

#### 0Dim = Sclar (consta val)
#### 1Dim = vector
#### 2Dim = matrix
#### 3Dim and more = tensor

```python
neuron = nn.Linear(in_features=3, out_features=1)
```

```python
output = neuron(inputs)
```

```python
output
```

```python
# To verify the Output -- formula
# Weighter Sum = Summation(XiWi + b)

neuron.weight @ inputs + neuron.bias      # @ --> multiplication
```
